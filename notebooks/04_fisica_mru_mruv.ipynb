# Física com Gráficos: MRU e MRUV

Nesta atividade vamos aprender com gráficos:

## MRU (Movimento Retilíneo Uniforme)
- velocidade constante
- posição cresce em linha reta

## MRUV (Movimento Retilíneo Uniformemente Variado)
- aceleração constante
- velocidade aumenta com o tempo
- posição vira uma curva


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 101)  # tempo de 0 a 10 segundos
```

```python
v0_mru = 5      # m/s
s0 = 0          # posição inicial

v_mru = np.full_like(t, v0_mru)
a_mru = np.zeros_like(t)
s_mru = s0 + v0_mru * t

df_mru = pd.DataFrame({"t(s)": t, "s_mru(m)": s_mru, "v_mru(m/s)": v_mru, "a_mru(m/s²)": a_mru})
df_mru.head()

```

```python
v0 = 2     # m/s
a = 1.5    # m/s²

v_mruv = v0 + a*t
a_mruv = np.full_like(t, a)
s_mruv = s0 + v0*t + 0.5*a*(t**2)

df_mruv = pd.DataFrame({"t(s)": t, "s_mruv(m)": s_mruv, "v_mruv(m/s)": v_mruv, "a_mruv(m/s²)": a_mruv})
df_mruv.head()
```

```python
plt.figure()
plt.plot(t, s_mru, label="MRU (posição)")
plt.plot(t, s_mruv, label="MRUV (posição)")
plt.title("Posição x Tempo")
plt.xlabel("Tempo (s)")
plt.ylabel("Posição (m)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(t, v_mru, label="MRU (velocidade)")
plt.plot(t, v_mruv, label="MRUV (velocidade)")
plt.title("Velocidade x Tempo")
plt.xlabel("Tempo (s)")
plt.ylabel("Velocidade (m/s)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(t, a_mru, label="MRU (aceleração)")
plt.plot(t, a_mruv, label="MRUV (aceleração)")
plt.title("Aceleração x Tempo")
plt.xlabel("Tempo (s)")
plt.ylabel("Aceleração (m/s²)")
plt.grid(True)
plt.legend()
plt.show()
```
