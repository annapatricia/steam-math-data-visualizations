# Funções Matemáticas e seus Gráficos

Nesta aula vamos visualizar:
- Função constante
- Função do 1º grau
- Função do 2º grau

Vamos entender como a equação vira um desenho no plano cartesiano.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.arange(-10, 11, 1)

y_const = np.full_like(x, 3)
y_1 = 2*x + 1
y_2 = x**2 - 4*x + 3

df = pd.DataFrame({
    "x": x,
    "Constante (3)": y_const,
    "1º grau (2x+1)": y_1,
    "2º grau (x²-4x+3)": y_2
})

df.head()
```

```python
plt.figure()
plt.plot(x, y_const, label="Constante")
plt.plot(x, y_1, label="1º grau")
plt.plot(x, y_2, label="2º grau")

plt.axhline(0)
plt.axvline(0)
plt.title("Comparação entre funções")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
