# População por Região do Brasil (Demografia)

Nesta atividade vamos:
- entender o que é **demografia** (estudo da população)
- comparar a população entre as regiões do Brasil
- construir gráficos para visualizar melhor os dados


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Região": ["Norte", "Nordeste", "Centro-Oeste", "Sudeste", "Sul"],
    "População_milhões": [18, 57, 16, 89, 30]
})

df
```

```python
plt.figure()
plt.bar(df["Região"], df["População_milhões"])
plt.title("População por Região do Brasil")
plt.xlabel("Região")
plt.ylabel("População (milhões)")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```

```python
plt.figure()
plt.pie(
    df["População_milhões"],
    labels=df["Região"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribuição Percentual da População (por Região)")
plt.show()
```

```python
df_sorted = df.sort_values("População_milhões", ascending=False).reset_index(drop=True)
df_sorted
```
